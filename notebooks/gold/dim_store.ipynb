```python
from pyspark.sql.functions import *
from delta.tables import DeltaTable
```

```python
silver_table = "retailnova.silver.stores"
gold_table = "retailnova.gold.dim_store"
```

```python
df = spark.table(silver_table).dropDuplicates(["store_id"])
```

```python
delta_table = DeltaTable.forName(spark, gold_table)
```

```python
delta_table.alias("target").merge(
    df.alias("source"),
    "target.store_id = source.store_id"
).whenMatchedUpdate(set={
    "store_name": "source.store_name",
    "city": "source.city",
    "state": "source.state",
    "region": "source.region",
    "store_type": "source.store_type",
    "opening_date": "source.opening_date"
}).whenNotMatchedInsert(values={
    "store_id": "source.store_id",
    "store_name": "source.store_name",
    "city": "source.city",
    "state": "source.state",
    "region": "source.region",
    "store_type": "source.store_type",
    "opening_date": "source.opening_date"
}).execute()
```

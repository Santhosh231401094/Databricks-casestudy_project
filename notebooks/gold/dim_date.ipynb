```python
from pyspark.sql.functions import *
```

```python
start_date = "2026-01-01"
end_date = "2026-12-31"
```

```python
df = spark.sql(f"""
    SELECT explode(
        sequence(
            to_date('{start_date}'),
            to_date('{end_date}'),
            interval 1 day
        )
    ) AS full_date
""")
```

```python
df = df.withColumn(
    "date_key",
    date_format(col("full_date"), "yyyyMMdd").cast("int")
)
```

```python
df = df.withColumn("year", year("full_date"))
df = df.withColumn("quarter", quarter("full_date"))
df = df.withColumn("month", month("full_date"))
df = df.withColumn("month_name", date_format("full_date", "MMMM"))
df = df.withColumn("day", dayofmonth("full_date"))
df = df.withColumn("day_of_week", dayofweek("full_date"))
df = df.withColumn("day_name", date_format("full_date", "EEEE"))
```

```python
df = df.select(
    "date_key",
    "full_date",
    "year",
    "quarter",
    "month",
    "month_name",
    "day",
    "day_of_week",
    "day_name"
)
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("retailnova.gold.dim_date")

print("dim_date created successfully")
```
